```python


from keras.models import load_model
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image

model = load_model("../models/facial_expression_model.h5")



image_path = "C:\\Users\\tanme\\OneDrive\\Desktop\\pet_emotion_visual\\notebooks\\dataset\\angry\\0C5yo7GxMy8lztxNZvSdfEx2gSPRTR701.jpg"


img = cv2.imread(image_path)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_resized = cv2.resize(img_gray, (48,48))
img_resized = img_resized.astype("float32") / 255.0

img_input = np.expand_dims(img_resized, axis=0)
img_input = np.expand_dims(img_input, axis=-1)



prediction = model.predict(img_input)
prediction_prob = prediction[0]

emotion_classes = {0:"happy",1:"sad",2:"angry"}

emotion_label = np.argmax(prediction_prob)
predicted_emotion = emotion_classes[emotion_label]



print("Predicted Emotion:", predicted_emotion)
print("Confidence [happy, sad, angry]:", prediction_prob)



plt.imshow(img_resized, cmap="gray")
plt.title(f"Predicted Emotion: {predicted_emotion}")
plt.axis("off")
plt.show()



img_pil = Image.open(image_path)
plt.imshow(np.array(img_pil))
plt.title(f"Predicted Emotion: {predicted_emotion}")
plt.axis("off")
plt.show()
```

```python
from keras.models import load_model
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image

model = load_model("../models/efficientnet_model.h5")

image_path = "C:\\Users\\tanme\\OneDrive\\Desktop\\pet_emotion_visual\\notebooks\\dataset\\angry\\0C5yo7GxMy8lztxNZvSdfEx2gSPRTR701.jpg"

# ── Preprocess image ──────────────────────────────────────
img = cv2.imread(image_path)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_resized = cv2.resize(img_gray, (48, 48))
img_resized = img_resized.astype("float32") / 255.0

# ── Fix grayscale → RGB ───────────────────────────────────
img_input = np.expand_dims(img_resized, axis=0)   # (1, 48, 48)
img_input = np.expand_dims(img_input, axis=-1)    # (1, 48, 48, 1)
img_input = np.repeat(img_input, 3, axis=-1)      # (1, 48, 48, 3) ← fix

# ── Predict ───────────────────────────────────────────────
prediction = model.predict(img_input)
prediction_prob = prediction[0]

emotion_classes = {0: "happy", 1: "sad", 2: "angry"}
emotion_label = np.argmax(prediction_prob)
predicted_emotion = emotion_classes[emotion_label]

print("Predicted Emotion:", predicted_emotion)
print("Confidence [happy, sad, angry]:", prediction_prob)

# ── Show grayscale image ──────────────────────────────────
plt.imshow(img_resized, cmap="gray")
plt.title(f"Predicted Emotion: {predicted_emotion}")
plt.axis("off")
plt.show()

# ── Show original color image ─────────────────────────────
img_pil = Image.open(image_path)
plt.imshow(np.array(img_pil))
plt.title(f"Predicted Emotion: {predicted_emotion}")
plt.axis("off")
plt.show()
```
