```python

import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50
from tensorflow.keras.applications.vgg16 import VGG16
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2
from tensorflow.keras.applications.efficientnet import EfficientNetB0
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model

def build_transfer_model(base):
    x=base.output
    x=GlobalAveragePooling2D()(x)
    x=Dense(256,activation='relu')(x)
    outputs=Dense(3,activation='softmax')(x)
    return Model(base.input,outputs)

base=ResNet50(weights='imagenet',include_top=False,input_shape=(224,224,3))
resnet_model=build_transfer_model(base)
resnet_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
resnet_model.summary()
```

```python
from tensorflow.keras.applications.vgg16 import VGG16
base = VGG16(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
vgg16_model = build_transfer_model(base)
vgg16_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
vgg16_model.summary()
```

```python
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2
mobilenet_model = build_transfer_model(base)
mobilenet_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mobilenet_model.summary()
```

```python
from tensorflow.keras.applications.efficientnet import EfficientNetB0
base = EfficientNetB0(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
efficientnet_model = build_transfer_model(base)
efficientnet_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
efficientnet_model.summary()
```

```python
import numpy as np
import cv2

# ── Load .npy files ───────────────────────────────────────
x_train = np.load(r"C:\Users\tanme\OneDrive\Desktop\pet_emotion_visual\processed_data\X_train.npy")
y_train = np.load(r"C:\Users\tanme\OneDrive\Desktop\pet_emotion_visual\processed_data\y_train.npy")
x_test  = np.load(r"C:\Users\tanme\OneDrive\Desktop\pet_emotion_visual\processed_data\X_test.npy")
y_test  = np.load(r"C:\Users\tanme\OneDrive\Desktop\pet_emotion_visual\processed_data\y_test.npy")

print("Before:", x_train.shape)

# ── Resize to 224x224 ─────────────────────────────────────
x_train = np.array([cv2.resize(img, (224, 224)) for img in x_train])
x_test  = np.array([cv2.resize(img, (224, 224)) for img in x_test])

# ── Fix grayscale → RGB ───────────────────────────────────
if x_train.ndim == 3:
    x_train = np.expand_dims(x_train, axis=-1)
    x_test  = np.expand_dims(x_test,  axis=-1)

if x_train.shape[-1] == 1:
    x_train = np.repeat(x_train, 3, axis=-1)
    x_test  = np.repeat(x_test,  3, axis=-1)

# ── Normalize ─────────────────────────────────────────────
x_train = x_train / 255.0
x_test  = x_test  / 255.0

print("After:", x_train.shape)  # Should be (samples, 224, 224, 3)

# ── Train all models ──────────────────────────────────────
history_resnet       = resnet_model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_test, y_test))
history_vgg16        = vgg16_model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_test, y_test))
history_mobilenet    = mobilenet_model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_test, y_test))
history_efficientnet = efficientnet_model.fit(x_train, y_train, epochs=10, batch_size=32, validation_data=(x_test, y_test))

# ── Evaluate ──────────────────────────────────────────────
loss_r, acc_r = resnet_model.evaluate(x_test, y_test)
loss_v, acc_v = vgg16_model.evaluate(x_test, y_test)
loss_m, acc_m = mobilenet_model.evaluate(x_test, y_test)
loss_e, acc_e = efficientnet_model.evaluate(x_test, y_test)

print(f"\nResNet50       → Accuracy: {acc_r*100:.2f}%  Loss: {loss_r:.4f}")
print(f"VGG16          → Accuracy: {acc_v*100:.2f}%  Loss: {loss_v:.4f}")
print(f"MobileNetV2    → Accuracy: {acc_m*100:.2f}%  Loss: {loss_m:.4f}")
print(f"EfficientNetB0 → Accuracy: {acc_e*100:.2f}%  Loss: {loss_e:.4f}")
```
